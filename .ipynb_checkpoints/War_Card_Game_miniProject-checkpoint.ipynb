```python
import random
suits=('Hearts','Diamonds','Spades','Clubs')
ranks=('Two','Three','Four','Five','Six','Seven','Eight','Nine','Ten','Jack','King','Queen','Ace')
values={'Two':2,'Three':3,'Four':4,'Five':5,'Six':6,'Seven':7,'Eight':8,'Nine':9,'Ten':10,'Jack':11,'King':12,'Queen':13,'Ace':14}
```

```python
class Card:
    def __init__(self,suits,rank):
        self.suit=suits
        self.rank=rank
        self.value=values[rank]

    def __str__(self):
        return self.rank + " Of " + self.suit
    
```

```python
two_heart=Card('Hearts','Two')
```

```python
print(two_heart)
```

```python
print (values[two_heart.rank])
```

```python
class Deck:
    
    def __init__(self):
        self.all_cards=[]
        for suit in suits:
            for rank in ranks:
                created_card=Card(suit,rank)
                self.all_cards.append(created_card)

    def shuffle(self):
        random.shuffle(self.all_cards)

    def deal_one(self):
        return self.all_cards.pop()
                
```

```python
new_deck=Deck()
```

```python
new_deck.shuffle()
```

```python
new_cards=new_deck.all_cards[0]
```

```python
my_card=new_deck.deal_one()
```

```python
print(my_card)
```
