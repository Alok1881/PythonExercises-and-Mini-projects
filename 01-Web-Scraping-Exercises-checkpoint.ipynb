**TASK: Notice how there is more than one page, and subsequent pages look like this http://quotes.toscrape.com/page/2/. Use what you know about for loops and string concatenation to loop through all the pages and get all the unique authors on the website. Keep in mind there are many ways to achieve this, also note that you will need to somehow figure out how to check that your loop is on the last page with quotes. For debugging purposes, I will let you know that there are only 10 pages, so the last page is http://quotes.toscrape.com/page/10/, but try to create a loop that is robust enough that it wouldn't matter to know the amount of pages beforehand, perhaps use try/except for this, its up to you!**

```python
# CODE HERE
import requests
from bs4 import BeautifulSoup

# Base URL setup for string concatenation
base_url = 'http://quotes.toscrape.com/page/'

# A set is perfect here because it only stores unique values
unique_authors = set()

# Initialize our page counter
page_number = 1

while True:
    # Concatenate the base URL with the current page number
    scrape_url = base_url + str(page_number) + '/'
    
    # Request the page
    res = requests.get(scrape_url)
    soup = BeautifulSoup(res.text, 'html.parser')
    
    # Select all HTML elements that contain the author's name
    # On this site, authors are stored in <small class="author"> tags
    authors_on_page = soup.select('.author')
    
    # THE BREAK CONDITION: 
    # If the list 'authors_on_page' is empty, it means we have reached a page 
    # (like page 11) that doesn't have any quotes. Time to stop looping!
    if not authors_on_page:
        print(f"Loop broken. Reached empty page at: {scrape_url}")
        break
        
    # If the page does have authors, loop through them and add to our set
    for author in authors_on_page:
        unique_authors.add(author.text)
        
    # Increment the page number to check the next one on the next loop iteration
    page_number += 1

print("\n--- Unique Authors Found ---")
for author in sorted(unique_authors):
    print(author)
```

There are lots of other potential solutions that are even more robust and flexible, the main idea is the same though, use a while loop to cycle through potential pages and have a break condition based on the invalid page.
