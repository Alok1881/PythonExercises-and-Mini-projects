# Simple Deposit Withdraw Banking Exercise

For this challenge, create a bank account class that has two attributes:

* owner
* balance

and two methods:

* deposit
* withdraw

As an added requirement, withdrawals may not exceed the available balance.

Instantiate your class, make several deposits and withdrawals, and test to make sure the account can't be overdrawn.

```python
class Account:
    def __init__(self,owner,balance):
        self.owner=owner
        self.balance=balance

    def __str__(self):
        return "Owner: %s, Balance: %s" %(self.owner, self.balance)

    def deposit(self,amount):
        self.amount=amount
        self.balance += amount
        return self.balance

    def withdraw(self,withdrawl_amount):
        self.withdrawl_amount=withdrawl_amount
        if withdrawl_amount>self.balance:
            print(f"Funds unavailable {self.balance}")
        else:
            self.balance -= withdrawl_amount
            print(f"Withdrawl accepted and new balance is {self.balance}")
```

```python
# 1. Instantiate the class
acct1 = Account('Jose',100)
```

```python
# 2. Print the object
print(acct1)
```

```python
# 4. Show the account balance attribute
acct1.balance
```

```python
# 5. Make a series of deposits and withdrawals
acct1.deposit(50)
```
