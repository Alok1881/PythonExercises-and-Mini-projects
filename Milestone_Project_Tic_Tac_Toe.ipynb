# Milestone Project 1
assignment: Create a Tic Tac Toe game. 

Here are the requirements:

* 2 players should be able to play the game (both sitting at the same computer)
* The board should be printed out every time a player makes a move
* You should be able to accept input of the player position and then place a symbol on the board

```python
def user_choice():
    choice='Wrong'
    acceptable_range= range(1,10)
    within_range=False
    while choice.isdigit()==False or within_range==False:
        choice=input(f"Enter number from 1-9:")
        if choice.isdigit()==False:
            print(f"Sorry,Please enter correct integer value")
        elif choice.isdigit()==True:
            if int(choice) in acceptable_range:
                within_range=True
            else:
                print(f"Number entered is out of range:")
                within_range=False
                
    return int(choice)
    
```

```python
def display_board(board):
    # Using f-strings just like you did, but mapped to a single list
    print(f" {board[7]} | {board[8]} | {board[9]} ")
    print("-----------")
    print(f" {board[4]} | {board[5]} | {board[6]} ")
    print("-----------")
    print(f" {board[1]} | {board[2]} | {board[3]} ")
```

```python
def game_on_choice():
    choice = 'Wrong'
    
    # Check if the choice is NOT in our list of valid answers
    while choice not in ['Y', 'N']:
        # .upper() converts inputs like 'y' to 'Y' automatically
        choice = input("Please enter your choice 'Y' or 'N': ").upper().strip()
        
        if choice not in ['Y', 'N']:
            print("Invalid input, please enter correct input.")
            
    # Once the loop breaks (meaning they entered Y or N), return it
    if choice == 'Y':
        return True
    else:
        return False

# Test it:
# is_game_on = game_on_choice()
```

```python
def enter_value():
    value = ' '
    while value not in ['O','X']:
        value = input("Enter either 'O' or 'X': ").upper().strip()
        if value not in ['O','X']:
            print("Invalid input, please enter either 'O' or 'X':")
    return value
```

```python
def check_win(board, mark):
    # Returns True if any of the 8 winning combinations are entirely filled by the given mark
    
    return (
        # Check Horizontal Rows
        (board[7] == mark and board[8] == mark and board[9] == mark) or # Top row
        (board[4] == mark and board[5] == mark and board[6] == mark) or # Middle row
        (board[1] == mark and board[2] == mark and board[3] == mark) or # Bottom row
        
        # Check Vertical Columns
        (board[7] == mark and board[4] == mark and board[1] == mark) or # Left column
        (board[8] == mark and board[5] == mark and board[2] == mark) or # Middle column
        (board[9] == mark and board[6] == mark and board[3] == mark) or # Right column
        
        # Check Diagonals
        (board[7] == mark and board[5] == mark and board[3] == mark) or # Diagonal 1
        (board[9] == mark and board[5] == mark and board[1] == mark)    # Diagonal 2
    )

# --- How to test it ---
# test_board = ['#', 'X', 'O', 'X', 'O', 'X', 'O', 'X', 'O', 'X']
# print(check_win(test_board, 'X'))  # This would return True (Diagonal win)
```

```python
def space_check(board, position):
    # Returns True if the requested position is empty
    return board[position] == ' '
```

```python
def full_board_check(board):
    # Loop through positions 1 to 9
    for i in range(1, 10):
        # If we find even one empty space, the board is NOT full
        if space_check(board, i):
            return False
            
    # If the loop finishes and finds no empty spaces, the board IS full
    return True
```

```python
import random
def choose_first():
    # Randomly choose which player goes first
    Flip = random.randint(0, 1)
    if     Flip == 0:
        return 'Player 1'
    else:
        return 'Player 2'
```

```python
def play_game():
    print("Welcome to Tic Tac Toe!")
    
    # 1. The Outer Loop (Allows for multiple games)
    while True:
        
        # --- SETUP PHASE ---
        the_board = [' '] * 10
        
        # Randomly choose which player goes first
        turn = choose_first()
        
        # Get Player 1's marker using your function
        player1_marker = enter_value()
        
        # Automatically assign the opposite marker to Player 2
        if player1_marker == 'X':
            player2_marker = 'O'
        else:
            player2_marker = 'X'
        
        # Print correct message about who goes first and their marker
        if turn == 'Player 1':
            print(f"{turn} will go first and play as {player1_marker}.")
        else:
            print(f"{turn} will go first and play as {player2_marker}.")
        
        # Ask if they are ready using your function
        game_is_running = game_on_choice()

        # --- THE GAME LOOP ---
        while game_is_running:
            
            if turn == 'Player 1':
                # Player 1's Turn
                display_board(the_board)
                print("\nPlayer 1's turn:")
                position = user_choice()
                
                # Check if space is valid before placing
                while not space_check(the_board, position):
                    print("That space is already taken! Choose another.")
                    position = user_choice()
                    
                the_board[position] = player1_marker
                
                # Check for win or tie
                if check_win(the_board, player1_marker):
                    display_board(the_board)
                    print("Congratulations! Player 1 has won the game!")
                    game_is_running = False
                elif full_board_check(the_board):
                    display_board(the_board)
                    print("The game is a tie!")
                    game_is_running = False
                else:
                    turn = 'Player 2' # Switch turn
                    
            else:
                # Player 2's Turn
                display_board(the_board)
                print("\nPlayer 2's turn:")
                position = user_choice()
                
                # Check if space is valid before placing
                while not space_check(the_board, position):
                    print("That space is already taken! Choose another.")
                    position = user_choice()
                    
                the_board[position] = player2_marker
                
                # Check for win or tie
                if check_win(the_board, player2_marker):
                    display_board(the_board)
                    print("Congratulations! Player 2 has won the game!")
                    game_is_running = False
                elif full_board_check(the_board):
                    display_board(the_board)
                    print("The game is a tie!")
                    game_is_running = False
                else:
                    turn = 'Player 1' # Switch turn back
                    
        # --- REPLAY PHASE ---
        print("\nWould you like to play another round?")
        if not game_on_choice():
            print("Thanks for playing! Goodbye.")
            break # Breaks the outer loop and ends the program

# Start the game!
play_game()
```
